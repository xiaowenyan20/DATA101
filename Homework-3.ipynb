```python
import numpy as np
import pandas as pd

titanic = pd.read_csv("titanic.csv")
titanic.info()
titanic.describe()

```

```python
mode_embarked = titanic['Embarked'].mode()
titanic['Embarked'] = titanic['Embarked'].fillna(mode_embarked)

mean_age = titanic['Age'].mean()
titanic['Age'] = titanic['Age'].fillna(mean_age)
```

```python
titanic['Age'] = titanic['Age'].fillna(titanic.groupby(['Sex', 'Pclass'])['Age'].transform('median'))
```

```python
titanic['FamilySize'] = titanic['SibSp'] + titanic['Parch'] +1

family_min= titanic['FamilySize'].min()
family_max= titanic['FamilySize'].max()
family_mean= titanic['FamilySize'].mean()

alone =( titanic['FamilySize'] == 1).sum()
print (alone)
```

```python
females = titanic[titanic['Sex'] == 'female']
males = titanic[titanic['Sex'] == 'male']
under16 = titanic[titanic['Age'] <16]

female_survival = females['Survived'].mean()
male_survival = males['Survived'].mean()
under16_survival = under16['Survived'].mean()

print(female_survival, male_survival, under16_survival)

```

```python
survival_rate = titanic.groupby(['Sex', 'Pclass'])['Survived'].mean()
print(survival_rate)
```

# Higher class passengers had a better chance of survival. 
# Overall, women and children had a much higher survival rate than men. 

```python
ticket_count = titanic.groupby("Ticket")["Ticket"].transform("count")
titanic['FarePerPerson'] = titanic['Fare']/ticket_count
adjusted = titanic.groupby('Pclass')['FarePerPerson'].mean()
print(adjusted)
```
