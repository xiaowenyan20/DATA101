```python
import numpy as np

students = []
tests = ["Homework", "Project", "Exam1", "Exam2", "Final Exam"]
grades = np.zeros((10,5))

for i in range(10):
    name = input(f"Please enter student name: ")
    students.append(name)

    print(f"Please enter scores for {name}: ")
    for j in range(5):
            score = float(input(f"{tests[j]}"))
            grades[i,j] = score

weight = np.array([0.15,0.15,0.20,0.20,0.30])

final_grade= grades @ weight

letter_grades= []

for score in final_grade:
    if score >= 90:
        letter_grades.append('A')
    elif score >= 80:
        letter_grades.append('B')
    elif score >= 70:
        letter_grades.append('C')
    elif score >= 60:
        letter_grades.append('D')
    else:
        letter_grades.append('F')
    
for i in range(10):
    print( students[i], ": " , final_grade[i], " ", letter_grades[i])
    
```
